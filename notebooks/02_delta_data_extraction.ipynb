```python
import pandas as pd

delta_data = pd.DataFrame([
    {
        "company": "Delta Corporation Limited",
        "report_year": 2026,
        "metric": "Revenue",
        "value": 1_091_805,
        "unit": "US$'000",
        "source": "Delta Corporation Limited Annual Report 2026",
    },
    {
        "company": "Delta Corporation Limited",
        "report_year": 2025,
        "metric": "Revenue",
        "value": 807_470,
        "unit": "US$'000",
        "source": "Delta Corporation Limited Annual Report 2026",
    },
])

delta_data

additional_metrics = pd.DataFrame([
    {
        "company": "Delta Corporation Limited",
        "report_year": 2026,
        "metric": "Net operating income",
        "value": 208_687,
        "unit": "US$'000",
        "source": "Delta Corporation Limited Annual Report 2026",
    },
    {
        "company": "Delta Corporation Limited",
        "report_year": 2025,
        "metric": "Net operating income",
        "value": 147_086,
        "unit": "US$'000",
        "source": "Delta Corporation Limited Annual Report 2026",
    },
    {
        "company": "Delta Corporation Limited",
        "report_year": 2026,
        "metric": "Total assets",
        "value": 591_435,
        "unit": "US$'000",
        "source": "Delta Corporation Limited Annual Report 2026",
    },
    {
        "company": "Delta Corporation Limited",
        "report_year": 2025,
        "metric": "Total assets",
        "value": 442_388,
        "unit": "US$'000",
        "source": "Delta Corporation Limited Annual Report 2026",
    },
])

delta_data = pd.concat([delta_data, additional_metrics], ignore_index=True)

delta_data
```

```python
print('Missing values per column:')
print(delta_data.isnull().sum())

print('\nDuplicate records:')
print(delta_data.duplicated().sum())

print('\nData types:')
print(delta_data.dtypes)
```

```python
#year on year revenue growth
revenue_2025 = delta_data.loc[
    (delta_data['metric'] == 'Revenue') & (delta_data['report_year'] == 2025),
    'value'
].iloc[0]

revenue_2026 = delta_data.loc[
    (delta_data['metric'] == 'Revenue') & (delta_data['report_year'] == 2026), 'value'
].iloc[0]

revenue_growth = (
(revenue_2026 - revenue_2025) / revenue_2025
)*100

print(f'Revenue growth: {revenue_growth:.2f}%')
```

```python
#net operating income growth
income_2025 = delta_data.loc[
    (delta_data['metric'] == 'Net operating income') & (delta_data['report_year'] == 2025),
    'value'
].iloc[0]

income_2026 = delta_data.loc[
    (delta_data['metric'] == 'Net operating income') & (delta_data['report_year'] == 2026),
    'value'
].iloc[0]

income_growth = (
    (income_2026 - income_2025) / income_2025
) * 100

print(f"Net operating income growth: {income_growth:.2f}%")
```

```python
#operating margin
margin_2025 = (income_2025 / revenue_2025) * 100
margin_2026 = (income_2026 / revenue_2026) * 100

print(f"2025 operating margin: {margin_2025:.2f}%")
print(f"2026 operating margin: {margin_2026:.2f}%")
print(f"Change in margin: {margin_2026 - margin_2025:.2f} percentage points")
```

```python
#asset growth rate
assets_2025 = delta_data.loc[
    (delta_data['metric'] == 'Total assets') & (delta_data['report_year'] == 2025),
    'value'
].iloc[0]

assets_2026 = delta_data.loc[
    (delta_data['metric'] == 'Total assets') & (delta_data['report_year'] == 2026), 'value'
].iloc[0]

asset_growth = ((assets_2026 - assets_2025) / assets_2025)*100

print(f'2025 total assets: US${assets_2025:,} thousand')
print(f'2026 total assets: US${assets_2026:,} thousand')
print(f'Total asset growth: {asset_growth:.2f}%')
```

```python
#return on assets(ROA): helps us assess how much net operating income the company generated relative to its assets.

print(income_2025, assets_2025, income_2026, assets_2026)

roa_2025 = (income_2025 / assets_2025) * 100
roa_2026 = (income_2026 / assets_2026) * 100
print(f'2025 return on assets: {roa_2025:.2f}%')
print(f'2026 return on assets: {roa_2026:.2f}%')
print(f'Change in ROA: {roa_2026 - roa_2025:.2f} percentage points')
```

```python
kpi_summary = pd.DataFrame({
    'Metric': ['Revenue growth',
               'Net operating income growth',
               'Operating margin (2025)',
               'Operating margin (2026)',
               'Total asset growth',
               'Operating income / year-end assets (2025)',
               'Operating income / year-end assets (2026)',],
    'Value': [
        revenue_growth,
        income_growth,
        margin_2025,
        margin_2026,
        asset_growth,
        roa_2025,
        roa_2026,
    ],
    'Unit': [
        '%',
        '%',
        '%',
        '%',
        '%',
        '%',
        '%',
    ],
})

kpi_summary.round(2)
```

## Reporting Basis and Comparability

The 2025 comparative figures were retrospectively restated following a reassessment of supplier rebate receivables, as described in Note 30 of Delta Corporation's 2026 Annual Report. The restatement reduced trade and other receivables by US$5.521 million, increased operating costs by US$5.521 million, and reduced retained earnings by US$4.099 million. The report states that the restatement had no net effect on prior-period cash flows.

Delta's functional currency changed to the US dollar effective 1 October 2023 and has been maintained. These reporting considerations must be taken into account when interpreting year-on-year changes.

**Source:** Delta Corporation Limited, *Annual Report 2026*, Note 30 and the Functional Currency disclosure.
